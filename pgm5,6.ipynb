6)Write a Python program to implement k-NN classification using the Iris dataset. Perform the
following steps:
1. Load the dataset using Pandas.
2. Separate the features and target variable.
3. Split the dataset into 80% training and 20% testing data.
4. Apply StandardScaler for feature scaling.
5. Train a k-NN classifier with k = 5.
6. Predict the classes of the test data.
7. Display the confusion matrix, classification report, and accuracy.





```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python

dataset=pd.read_csv("/content/Iris - Iris.csv")
print(dataset.head())
```

```python
x=dataset.iloc[:, :-1].values
y=dataset.iloc[:, 5].values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=0)
```

```python
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
classifier=KNeighborsClassifier(n_neighbors=5)
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("Classification Report:")
print(classification_report(y_test, y_pred))
```
