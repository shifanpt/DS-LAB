```python
#2
year = [2001, 2002, 2003, 2004, 2005, 2006, 2007]
carvalue = [24000, 22500, 19800, 17500, 14500, 10000, 5800]
plt.plot(year,carvalue,color='red',linestyle='-.',marker='*',markersize=20,markerfacecolor='green')
plt.title("Value depreciation(left aligned)")
plt.xlabel("Year")
plt.ylabel("Car value")
plt.legend()
plt.show()
```

```python
#3
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
Affordable = [173,153,195,147,120,144,148,109,174,130,172,131]
Luxury = [189,189,105,112,173,109,151,197,174,145,177,161]
Super_luxury = [185,185,126,134,196,153,112,133,200,145,167,110]
plt.scatter(months, Affordable, color='pink', label='Affordable Segment')
plt.scatter(months, Luxury, color='yellow', label='Luxury Segment')
plt.scatter(months, Super_luxury, color='blue', label='Super Luxury Segment')
plt.title("Sales Data")
plt.xlabel("Months of Year", fontsize=18)
plt.ylabel("Sales of Segments")
plt.legend()
plt.show()
```

```python
#4-(a)
import pandas as pd
import matplotlib.pyplot as plt
iris=pd.read_csv("/content/Iris - Iris.csv")
print(iris.head())
```

```python
#(b)
iris = pd.read_csv("/content/Iris - Iris.csv")
plt.scatter(iris["SepalLengthCm"], iris["SepalWidthCm"], color="blue")
plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
#(c)
colors = {"Iris-setosa":"red","Iris-versicolor":"green","Iris-virginica":"blue"}

for flower in colors:
    data = iris[iris["Species"] == flower]
    plt.scatter(data["SepalLengthCm"],data["SepalWidthCm"],color=colors[flower],label=flower)
plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.show()
```

```python
#(d)
iris = pd.read_csv("/content/Iris - Iris.csv")

plt.plot(iris["SepalLengthCm"], iris["SepalWidthCm"], color="red")
plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
#(e)
plt.hist(iris["SepalLengthCm"],color="orange",edgecolor="black")
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
#(f)
total=iris.groupby("Species")["SepalLengthCm"].sum()
plt.bar(total.index,total.values,color=["red","green","blue"])
plt.title("Total Sepal Length")
plt.xlabel("Species")
plt.ylabel("Total Sepal Length")
plt.show()
```

```python
#(g)
data = [
    iris[iris["Species"]=="Iris-setosa"]["SepalLengthCm"],
    iris[iris["Species"]=="Iris-versicolor"]["SepalLengthCm"],
    iris[iris["Species"]=="Iris-virginica"]["SepalLengthCm"]
]

plt.boxplot(
    data,
    labels=["Setosa","Versicolor","Virginica"])

plt.title("Sepal Length Distribution")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()
```

```python
#(h)
count=iris["Species"].value_counts()
plt.pie(count,labels=count.index,autopct="%1.1f%%")
plt.title("Species Distribution")
plt.show()
```

```python
#(i)
colors = {"Iris-setosa":"red","Iris-versicolor":"green","Iris-virginica":"blue"}

for flower in colors:
    data = iris[iris["Species"]==flower]

    plt.scatter(data["PetalLengthCm"],
                data["PetalWidthCm"],
                color=colors[flower],
                label=flower)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.show()
```

```python
#(j)
plt.subplot(2,2,1)
plt.hist(iris["SepalLengthCm"])
plt.title("Sepal Length")

plt.subplot(2,2,2)
plt.hist(iris["SepalWidthCm"])
plt.title("Sepal Width")

plt.subplot(2,2,3)
plt.hist(iris["PetalLengthCm"])
plt.title("Petal Length")

plt.subplot(2,2,4)
plt.hist(iris["PetalWidthCm"])
plt.title("Petal Width")

plt.tight_layout()
plt.show()
```

```python
#(k)
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

iris = pd.read_csv("/content/Iris - Iris.csv")

avg = iris.groupby("Species")[["SepalLengthCm","SepalWidthCm"]].mean()

x = np.arange(len(avg.index))
width = 0.35

plt.bar(x - width/2, avg["SepalLengthCm"], width, label="Sepal Length")
plt.bar(x + width/2, avg["SepalWidthCm"], width, label="Sepal Width")

plt.xticks(x, avg.index)
plt.xlabel("Species")
plt.ylabel("Average")
plt.title("Average Sepal Length and Sepal Width")
plt.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/Iris - Iris.csv")

plt.figure(figsize=(8, 6))

plt.scatter(
    df["PetalLengthCm"],
    df["PetalWidthCm"],
    s=df["SepalLengthCm"] * 20,
    alpha=0.6
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Bubble Scatter Plot")
plt.show()
```
